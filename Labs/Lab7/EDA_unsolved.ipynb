```python
# ============================================================
# LAB: Exploratory Data Analysis (Student Version)
# Dataset: FIFA Players
# Goal: Discover insights by exploring the data yourself
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("seaborn-v0_8")
```

```python
# ==============================
# 2. Load Dataset
# ==============================

# TODO: Replace with your dataset
df = pd.read_csv("data/StudentsPerformance.csv")

# TODO: Show basic information
df.info()
```

```python
# TODO:
# Create a new column called "average_score"
df['average_score'] = (df['math score'] + df['reading score'] + df['writing score']) / 3
df['average_score'].head()
```

Q: how is the distribution of the new score?

```python
# TODO:
# Plot the distribution of the average score
col = "average_score"
plt.figure(figsize=(10, 6))
plt.hist(df[col], bins=60)
plt.title("average_score Distribution")
plt.xlabel("average_score")
plt.ylabel("Number of Students")
plt.show()
```

Q1: Do students who completed the test preparation course perform better?

```python
# TODO: which plot do you choose?
sns.boxplot(data=df, x="test preparation course", y="average_score") 
plt.title("Correlation between test preparation course and average_score")
plt.xlabel("test preparation course")
plt.ylabel("average_score")
plt.show()
```

Q2: Does lunch type affect performance?

```python
# TODO:
sns.boxplot(data=df, x="lunch", y="average_score") 
plt.title("Correlation between lunch and average_score")
plt.xlabel("lunch")
plt.ylabel("average_score")
plt.show()
```

Q3: do the three features imapct each other? 
Gender + Test Preparation + Performance

```python
# TODO:
sns.boxplot(data=df, x="test preparation course", y="average_score", hue='gender') 
plt.title("Correlation between test preparation course and average_score")
plt.xlabel("test preparation course")
plt.ylabel("average_score")
plt.show()
```

Q4: Parental Education + Gender + Performance

```python
# TODO:
sns.boxplot(data=df, x="gender", y="average_score", hue='parental level of education') 
plt.title("Correlation between parental level of education and average_score and gender")
plt.xlabel("gender")
plt.ylabel("average_score")
plt.show()
```
