```python
import pandas as pd
import numpy as np
from parametrictests import ParametricTests

data = pd.read_csv("StudentsPerformance.csv")
#adding the average score column to the data frame
data["average_score"] = (
    data["math score"] +
    data["reading score"] +
    data["writing score"]
) / 3
```

## Q1. Is the average math score significantly different from 70?
**Compare ONE Sample Mean to a Target Value** 

```python
#TODO.1
# Create an instance of the ParametricTests class
# Run the test
# make decision

pt = ParametricTests()


mu0 = 70

t_stat, df = pt.one_sample_t_stat(data["math score"], mu0)

p_value, reject = pt.make_decision(t_stat, df, tailing = "two_tailed", alpha=0.05) # ---this is a dummy variable, replace it with the actual result from the test --- IGNORE ---
print(f"t-statistic: {t_stat}, degrees of freedom: {df}, p_value: {p_value}")

if reject:
    print("Reject the null hypothesis: the average math score is significantly different from 70")
else:
    print("Fail to reject the null hypothesis: the average math score is NOT significantly different from 70")
```

## Q2. Do students who completed the test-preparation course perform better with their scores?
**Compare means of TWO INDEPENDENT Groups** 

```python
#TODO.2
# get the required groups
# Run the test
# make decision



data1 = data.loc[data['test preparation course'] == 'none', 'math score']
data2 = data.loc[data['test preparation course'] == 'completed', 'math score']

t_stat, df = pt.two_sample_t_stat(data1, data2)

p_value, reject = pt.make_decision(t_stat, df, tailing = "two_tailed", alpha=0.05)

print(f"t-statistic: {t_stat}, degrees of freedom: {df}, p_value: {p_value}")

if reject:
    print("Reject the null hypothesis: yes, the students who completed the test preparation course have significantly different scores than those who did not complete the course")
else:   
    print("Fail to reject the null hypothesis: the completeness of the test preparation course has NO effect on the math scores of the students")
```

## Q3. Is there a significant difference between reading and writing scores for the same students?
**Compare scores for the SAME GROUP (DEPENDENT)** 


Remember: Dependency is not only for before/after. It is when the two values are linked to the same object.

```python
#TODO.3
# Run the test
# make decision



t_stat, df = pt.paired_t_stat(data['reading score'], data['writing score'])


p_value, reject = pt.make_decision(t_stat, df, tailing = "two_tailed", alpha=0.05)

print(f"t-statistic: {t_stat}, degrees of freedom: {df}, p_value: {p_value}")

if reject:
    print("Reject the null hypothesis: reading and writing scores are significantly different for the same students")
else:   
    print("Fail to reject the null hypothesis: reading and writing scores are NOT significantly different for the same students")
```

## Q4. Are males and females equally represented in the dataset?
**Testing if one variable matches a pattern**

```python
# TODO.4
# 1) get the observed and expected frequencies

obs_series = data['gender'].value_counts()
observed = obs_series.values  

# 2) Calculate Expected Counts (E) ---
# If we expect a 50/50 split, each category should be (Total / 2).
total_n = len(data)
expected = np.array([total_n / 2, total_n / 2])

# 3) Run the Test ---

stat, df, p = pt.chi_goodness_fit(observed, expected)

print(f"Observed Counts: {obs_series.to_dict()}")
print(f"P-value: {p:.4f}")



reject = p < 0.05

if reject:
    print("Result is significant (Reject H0): The distribution is NOT what we expected.")
else:
    print("Result is not significant (Keep H0): The data matches our expectation.")
```

## Q5. What feature is related to the total score? Lunch, Test preparation completeness, or parental level of education 
**Are two categorical variables associated or linked?**

> this is the same analysis of the previous lab, but now we answer them with statistics


```python
# TODO.4 "lunch"
# Run the Test ---
# Make decision

data1 = data.loc[data['lunch'] == 'free/reduced', 'average_score']
data2 = data.loc[data['lunch'] == 'standard', 'average_score']

t_stat, df = pt.two_sample_t_stat(data1, data2)

p_value, reject = pt.make_decision(t_stat, df, tailing = "two_tailed", alpha=0.05)

print(f"t-statistic: {t_stat}, degrees of freedom: {df}, p_value: {p_value}")

if reject:
    print("Result is significant (Reject H0): Lunch affects the student score.")
else:
    print("Result is not significant (Keep H0): Lunch is NOT linked to the student score.")
```
