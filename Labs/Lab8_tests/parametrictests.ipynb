```python
import pandas as pd

# Creating your example dataset
data = {
    'feature1': ['M', 'F', 'M', 'F', 'F', 'M', 'M', 'F'],
    'feature2': ['Complete', 'Complete', 'Incomplete', 
                 'Complete', 'Incomplete', 'Incomplete', 'Complete', 'Complete']
}
df = pd.DataFrame(data)

# --- STEP 1: Get Observed Counts (O) ---
# We count how many 'Complete' vs 'Incomplete' results are in the column.
obs_series = df['feature2'].value_counts()
observed = obs_series.values  # e.g., [5, 3]

# --- STEP 2: Calculate Expected Counts (E) ---
# If we expect a 50/50 split, each category should be (Total / 2).
total_n = len(df)
expected = np.array([total_n / 2, total_n / 2]) # e.g., [4, 4]

# --- STEP 3: Run the Test ---
stat, df, p, reject = pt.chi_goodness_fit(observed, expected)

print(f"Observed Counts: {obs_series.to_dict()}")
print(f"P-value: {p:.4f}")

if reject:
    print("Result is significant (Reject H0): The distribution is NOT what we expected.")
else:
    print("Result is not significant (Keep H0): The data matches our expectation.")
```
